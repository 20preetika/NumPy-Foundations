#ITERATING

```python
a6
for i in a6:
    print(i)
```

```python
a7                  #in 2-D when we use loop, for one iteration it displays 1 row.
for i in a7:
    print(i)
```

```python
a8                 #in 3-D when we use loop, for one iteration it displays 1 2-D array.
for i in a8:
    print(i)
```

```python
##in order to run loop on each item in 2-D and 3-D array , we use a func called nditer() 
##which converts any dimension array into 1-D array and then prints it.
for i in np.nditer(a8):
    print(i)
```

#STACKING usecase - when we have multiple datasources, so we bring data from various sources and stack them , we just need shape should be same.

```python
#horizontal stacking
a9 = np.arange(12).reshape(3,4)
a10 = np.arange(12,24).reshape(3,4)
a10
```

```python
np.hstack((a9,a10))
```

```python
#vertical stacking
```

```python
np.vstack((a9,a10))
```

#SPLITING - opp of stacking, when we want to split data

```python
#horizontal splitting
a9
np.hsplit(a9,2)
```

```python
#vertical splitting
a10
```

```python
np.vsplit(a10,3)
```
