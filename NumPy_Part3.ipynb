```python
#np.concatenate → It joins multiple arrays together along a specified axis.
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)
```

Key point:
axis=0 → stack rows (vertical)
axis=1 → stack columns (horizontal)

```python
np.concatenate((c,d),axis = 0)
```

```python
np.concatenate((c,d),axis = 1)
```

#SET FUNCTIONS np.union1d, np.intersect1d, np.setdiff1d, np.setxor1d, np.in1d

```python
m = np.array([1,2,4,5,6])
n = np.array([3,5,8,9,4])

np.union1d(m, n)
```

```python
np.intersect1d(m, n)
```

```python
np.setdiff1d(m, n)   #items in m but not in n 
```

```python
np.setxor1d(m, n)  #remove commons and keep rest
```

```python
np.in1d(m, 4)  #np.in1d returns a boolean array showing membership of elements.
```
